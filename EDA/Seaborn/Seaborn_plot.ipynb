```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset("flights")
```

```python
flights.head()
```

```python
x=flights.pivot_table(index="year", columns="month", values="passengers",aggfunc="sum")
```

```python
sns.heatmap(x)
```

```python
sns.heatmap(x,cbar=False)
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5)
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True, fmt='d')
```

```python
sns.heatmap(x,cbar=False,linewidths=0.5,annot=True, fmt='d',cmap='coolwarm')
```

```python
# Cluster map

sns.clustermap(x)
```

```python
sns.clustermap(x,metric='correlation')
```

```python
sns.clustermap(x,z_score=0,metric='correlation')
```

```python

sns.clustermap(x,z_score=0,annot=True,metric='correlation')
```

```python
sns.clustermap(x,z_score=0,annot=True,row_cluster=False,metric='correlation')
```

```python
sns.clustermap(x,z_score=0,annot=True,col_cluster=False,metric='correlation')
```

```python
# Joint Plot

tips = sns.load_dataset("tips")
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips)
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips,kind="hex")
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips,kind="kde")
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips,kind="reg")
```

```python
# Distplots

titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic['age'].fillna(titanic['age'].mean(),inplace=True)
```

```python
sns.distplot(titanic['age'])
```

```python
sns.distplot(titanic['age'],kde=False)
```

```python
sns.distplot(titanic['age'],kde=False,bins=100)
```

```python
sns.distplot(titanic['age'],hist=False)
```

```python
sns.distplot(titanic['age'],kde=False,hist=False,rug=True)
```

```python
sns.distplot(titanic[titanic['survived']==1]['age'])
sns.distplot(titanic[titanic['survived']==0]['age'])
```
