```python
batsman = pd.read_csv("sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# multiple plot
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# label titles
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title("Rohit Sharma vs Virat Kohli Carrer Comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# colors and line and marker
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red')

plt.title("Rohit Sharma vs Virat Kohli Carrer Comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
# color = hexcode and line style=dashed,dotted,dashdot,linewidth=5,3,6,4
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dotted')

plt.title("Rohit Sharma vs Virat Kohli Carrer Comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dashed',marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dotted',marker='+',markersize=10)

plt.title("Rohit Sharma vs Virat Kohli Carrer Comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dashed',marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dotted',marker='+',markersize=10,label='Sharma')

plt.title("Rohit Sharma vs Virat Kohli Carrer Comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend()

plt.grid()
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='#D9F10F',linestyle='dashed',marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='#FC00D6',linestyle='dotted',marker='+',markersize=10,label='Sharma')

plt.title("Rohit Sharma vs Virat Kohli Carrer Comparison")
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend()

plt.grid()
plt.show()
```

```python
df = pd.read_csv("batsman_season_record.csv")
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
```

```python
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
children =[10,20,40,10,30]
colors=['red red red red red','blue blueblueblue','green eeneeneen','yeloow','pink'] 

plt.bar(colors,children,color='black')
```

```python
# Stack bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=df['2016']+df['2017'],label='2015')

plt.legend()
plt.show()
```
